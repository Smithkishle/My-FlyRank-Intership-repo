## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

**Task Type:** Ranking / Scoring

**Why:** We are trying to answer the question "Which pages should we review first?". We aren't predicting a simple yes/no outcome, nor are we clustering unknown groups. We want to assign a priority score (opportunity gap) to each visible page and rank them so reviewers work on the highest-yield pages first.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

**Target:** A priority score representing the CTR opportunity gap (expected CTR for the position tier minus the page's actual CTR).

This target is **observed** indirectly, derived from actual impression and click data over the last 90 days, rather than a human-defined rule. We are modeling the expected relationship between position and CTR to score the deviation.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

**Success Metric:** Precision@K (e.g., Precision@100).

If we hand the top 100 ranked pages to a reviewer, what percentage of them are *actual* opportunities (i.e., truly underperforming their tier average with statistically significant volume, rather than just noise)? A higher precision means less wasted reviewer time.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Expected CTR varies wildly and non-linearly with search position. A simple rule like "flag anything under 2% CTR" would mistakenly flag pages that rank well but are in low-CTR SERP features, and completely miss pages ranking #1 that should have 15% CTR but only have 5%. We need to compare a page against its specific cohort (position tier) and factor in impression volume confidence, which is a scoring problem too complex for a clean if-statement.
